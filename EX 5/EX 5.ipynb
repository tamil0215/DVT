```python
import matplotlib.pyplot as plt

months = ["Jan","Feb","Mar","Apr","May","Jun"]
sales = [120,150,180,170,220,250]

plt.plot(months, sales, marker='o')
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
plt.bar(months, sales, color='orange')
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
departments = ["HR","IT","Finance","Sales"]
employees = [25,40,20,15]

plt.pie(employees, labels=departments, autopct='%1.1f%%', startangle=90)
plt.title("Employee Distribution")
plt.show()
```

```python
marks = [45,55,60,70,75,80,85,90,95,100,65,70,75,80,85]

plt.hist(marks, bins=5, color='skyblue', edgecolor='black')
plt.title("Student Marks Distribution")
plt.xlabel("Marks Range")
plt.ylabel("Frequency")
plt.show()
```

```python
hours = [2,3,4,5,6,7,8,9,10]
marks = [50,55,60,65,70,75,80,85,90]

plt.scatter(hours, marks, color='red')
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

```python
salary = [25000,27000,30000,32000,35000,40000,150000]

plt.boxplot(salary)
plt.title("Salary Distribution")
plt.show()
```

```python
months = ["Jan","Feb","Mar","Apr"]
sales = [120,150,180,200]
profit = [20,30,40,50]

plt.plot(months, sales, label="Sales", marker='o')
plt.plot(months, profit, label="Profit", marker='s')
plt.title("Sales vs Profit Trend")
plt.xlabel("Months")
plt.ylabel("Values")
plt.legend()
plt.show()
```
