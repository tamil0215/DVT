```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("indexData.csv")

print(df.head())
```

```python
print(df.shape)
```

```python
print(df.dtypes)
```

```python
high_close = df[df["Close"] > df["Close"].mean()]

print(high_close)
```

```python
high_volume = df[df["Volume"] > df["Volume"].mean()]

print(high_volume)
```

```python
sorted_df = df.sort_values(
    by="Close",
    ascending=False
)

print(sorted_df)
```

```python
print("\nStatistical Analysis:")

print("Mean =", df["Close"].mean())
print("Median =", df["Close"].median())
print("Mode =", df["Close"].mode().values)
print("Standard Deviation =", df["Close"].std())
```

```python
print("\nDescriptive Statistics:")
print(df.describe())
```

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df["Year"] = df["Date"].dt.year
year_avg = df.groupby("Year")["Close"].mean()

print("\nYear-wise Average Closing Price:")
print(year_avg)
```

```python
corr = df[
    ["Open", "High", "Low", "Close", "Adj Close", "Volume"]
].corr()

print("\nCorrelation Matrix:")
print(corr)
```

```python
plt.figure(figsize=(10, 5))

plt.plot(df["Date"], df["Close"])

plt.title("Stock Closing Price Analysis")
plt.xlabel("Date")
plt.ylabel("Close Price")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.bar(year_avg.index, year_avg.values)

plt.title("Year-wise Average Closing Price")
plt.xlabel("Year")
plt.ylabel("Average Close Price")

plt.show()
```
