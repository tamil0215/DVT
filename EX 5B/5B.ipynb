```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("indexData.csv")

print("Shape:", df.shape)


# ==========================================
# 1. INDEX-WISE RECORD COUNT
# ==========================================

index_count = df["Index"].value_counts()

plt.figure(figsize=(10,5))
plt.plot(index_count.index, index_count.values, marker='o')

plt.title("Number of Records by Index")
plt.xlabel("Index")
plt.ylabel("Number of Records")

plt.xticks(rotation=45)
plt.grid(True)
plt.show()


# ==========================================
# 2. RECORDS BY INDEX
# ==========================================

index_data = df["Index"].value_counts()

plt.figure(figsize=(10,5))
plt.bar(index_data.index, index_data.values)

plt.title("Records by Index")
plt.xlabel("Index")
plt.ylabel("Number of Records")

plt.xticks(rotation=45)
plt.show()


# ==========================================
# 3. INDEX DISTRIBUTION
# ==========================================

index_data = df["Index"].value_counts()

plt.figure(figsize=(8,6))

plt.pie(
    index_data.values,
    labels=index_data.index,
    autopct='%1.1f%%'
)

plt.title("Index Distribution")
plt.show()


# ==========================================
# 4. DISTRIBUTION OF CLOSE PRICE
# ==========================================

plt.figure(figsize=(8,5))

plt.hist(df["Close"].dropna(), bins=30)

plt.title("Distribution of Closing Price")
plt.xlabel("Close Price")
plt.ylabel("Frequency")

plt.show()


# ==========================================
# 5. OPEN PRICE VS CLOSE PRICE
# ==========================================

plt.figure(figsize=(8,5))

plt.scatter(
    df["Open"],
    df["Close"],
    alpha=0.3
)

plt.title("Open Price vs Close Price")
plt.xlabel("Open Price")
plt.ylabel("Close Price")

plt.show()


# ==========================================
# 6. CLOSE PRICE BOX PLOT
# ==========================================

plt.figure(figsize=(7,5))

plt.boxplot(df["Close"].dropna())

plt.title("Closing Price Box Plot")
plt.ylabel("Close Price")

plt.show()


# ==========================================
# 7. BASIC DATA ANALYSIS
# ==========================================

print("Total Records:", len(df))

print("Average Open Price:",
      round(df["Open"].mean(), 2))

print("Average Close Price:",
      round(df["Close"].mean(), 2))

print("Highest Close Price:",
      round(df["Close"].max(), 2))

print("Lowest Close Price:",
      round(df["Close"].min(), 2))

print("Most Common Index:",
      df["Index"].mode()[0])
```
