```python
import sqlite3
import pandas as pd
df = pd.read_csv("indexData.csv")
conn = sqlite3.connect("indexData.db")
df.to_sql("index_table", conn, if_exists="replace", index=False)
query = "SELECT * FROM index_table"

sql_df = pd.read_sql(query, conn)

print(sql_df.head())
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("indexData.csv")

index_count = df["Index"].value_counts()

index_count.plot(kind="bar")

plt.title("Number of Records for Each Stock Index")
plt.xlabel("Stock Index")
plt.ylabel("Count")

plt.show()
```
